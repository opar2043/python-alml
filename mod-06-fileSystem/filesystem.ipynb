### Read File

```python
with open("file.txt" , "r") as file :
    data = file.read()
    print(data)
    # print(file.readline())
    # print(file.readlines())
```

### Append

```python
with open("file.txt" , "a") as f:
    data = f.write("\nDeveloping my skills")
print(data)    
```

### File Pointer

```python
with open("file.txt", "r") as file:
    print(file.tell())  # Print the current position of the file pointer
    print(file.read())  # Read and print the contents of the file
    print(file.read(5))  # Read and print the first 15 characters of the file
    print(file.tell())  # Print the current position of the file pointer

    print(file.seek(0))  # Move the file pointer to the beginning of the file
    print(file.read(20))  # Read and print the first 5 characters of the file
```
