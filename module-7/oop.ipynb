```python
class Student:
    def __init__(self , name , age):
        self.name = name
        self.age = age

    def show(self): 
        print(f"Name: {self.name}, Age: {self.age}")

class Professional(Student):
    def __init__(self, name, age):
        super().__init__(name, age)

    def job(self , job):
        print(f"{self.name} works as a {job}")


s1= Student("Alice", 20)
s1.show()
p1 = Professional("Opar" , 25)
p1.job("Engineer")
p1.show()
```

OOP	Programming with classes and objects
Class	Blueprint for objects
Object	Instance of a class
Inheritance	Reusing another class's features
Polymorphism	Same interface, different behavior
Encapsulation	Bundling data and methods, controlling access
Abstraction	Hiding implementation details
Module	A Python file containing reusable code

```python
class Car:
    def run(self):
        print("car is running")

    def color(self, color):
        print(f"This is {color} color")

    def food(self , food , day , kg):
        print(f"This is {food} food for {day} days, {kg} kg")

c = Car()
c.run()
c.color("Blue")
c.food("Burger", 1, 6)

```

```python
class Mobile :
    def __init__(self, brand, model):
        self._brand = brand
        self._model = model

    def show(self):
        print(f"Brand Name {self._brand} and Model Name {self._model}")  

m1 = Mobile("Apple", "iPhone 12")
m1.show()
m1._brand = "Samsung"
m1.show()
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        print(self.side * self.side)

s = Square(4)
s.area() 
```

```python
from abc  import ABC , abstractmethod

# Abstruc 
class Bank(ABC):
    @abstractmethod
    def  show_balance(self):
        pass

# Encapsulation
class Account:
    def __init__(self , money):
        self._money = money

    def deposite(self, amount):
        print(f"Depositing {amount}")
        self._money = amount + self._money
        print(f"New balance: {self._money}")

    def withdraw (self, amount):
        if self._money >= amount:
            self._money = self._money - amount
            print(f"Withdrawing {amount}")

        else:
            print("Insufficient funds")

    def show_balance(self):
        print(f"Current balance: {self._money}")

# polymorphisum
 
class SavingAccount(Account):
      pass


opar = SavingAccount(500)
opar.show_balance()

opar.deposite(600)
opar.withdraw(77)
opar.show_balance()
```
