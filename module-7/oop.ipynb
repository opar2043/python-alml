```python
class Car:
    def run(self):
        print("car is running")

    def color(self, color):
        print(f"This is {color} color")

    def food(self , food , day , kg):
        print(f"This is {food} food for {day} days, {kg} kg")

c = Car()
c.run()
c.color("Blue")
c.food("Burger", 1, 6)
```
