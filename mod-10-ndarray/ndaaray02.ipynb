```python
# dictionary 
import numpy as np
dc = {"a": 5 , "b": 10 , "c": 15}
keys = dc.keys()
values = dc.values()
items = dc.items()

print(keys)
print(values)
print(items)

arr = np.array(list(values))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# items of ndarray 
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

### creating nd array from scratch

```python
arr = np.zeros((2, 2))
print(arr)
print(type(arr))
# print(arr.dtype)
# print(arr.shape)
# print(arr.ndim)
# print(arr.size)

arr3 = np.array([[[1, 2], [3, 4]] , [[5, 6], [7, 8]]])
print(arr3.ndim)

arr1 = np.zeros_like(arr3)
print(arr1)
```

```python
# np ones
arr = np.ones((2, 2) , dtype=np.int16)
print(arr)

arr = np.ones_like(arr3)
print(arr.shape)
print(arr)   
```

```python
# np empty 

arr = np.empty((4, 3) , dtype=np.int8)
print(arr)

arr = np.empty_like(arr3)
print(arr)   

# np fulll
arr = np.full((2, 2), np.inf)
print(arr)

arr = np.full_like(arr3,np.inf,dtype=np.float64)

print(arr)
print(arr.shape)
print(arr.dtype)
```
