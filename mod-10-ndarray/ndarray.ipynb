### numpy - numerical py 

```python
arr =[1,2,3]
print(arr * 5)

arr = [ x*3 for x in arr]
print(arr)
```

```python
import numpy as np
arr1 = np.array(arr)
print(arr1 * 5)
```

```python
import numpy as np 

# 1D array 
arr1 = np.array([1,2,3,4,5])
# 2D array 
arr2 = np.array([[1,2,3,4] , [10,20,30,40] , [5,6,7,8]])
# 3D array
arr3 = np.array([[[1,2,3,4] , [10,20,30,40]], [[5,6,7,8] , [50,60,70,80]]])

# array attributes 

print(arr1.ndim)
# print(type(arr1))
print(arr2.ndim)
print(arr3.ndim)

print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

print(arr1.size)
print(arr2.size)
print(arr3.size)
```

```python
flt= np.array([1.2, 2.3])
print(flt.dtype)

#  upcasting 
arrf = np.array([1, 2, 3 , 66.4])
print(arrf.dtype)
arrs = np.array([1, 2, 3 , 'hello'  ])
print(arrs.dtype)
```

```python
arr = np.array([1,2,3] , dtype= np.int16)
print(arr)
print(arr.dtype)
arr32 = arr.astype(np.int32)
print(arr32.dtype)
arr = np.array([1,2,3.2,5.6],dtype=np.int32)
print(arr.dtype)
```

```python
# ndarray creation


lst = [1, 2, 3, 4, 5.8]
arr = np.array(lst , dtype= np.int32)
print(arr)
print(type(arr))
print(arr.dtype)

matrix = [[1, 2, 3], [4, 5, 6]]
arr_matrix = np.array(matrix)
print(arr_matrix)
print(arr_matrix.dtype)
print(arr_matrix.size)
print(arr_matrix.shape)
print(arr_matrix.ndim)
```

```python
#  tuple 
tpl = (1, 2, 3, 4, 5)
arr = np.array(tpl , dtype=np.int32)
print(tpl)
print(arr)
print(type(arr))
print(arr.dtype)
arr = arr.astype(dtype=np.float64)
print(arr)
print(arr.dtype)
```

```python
# set 

st = {1, 2, 3, 4, 5}
arrs = np.array(list(st) , dtype=np.int32)
arrs = np.array(list(st))
print(arrs)
print(type(arrs))
print(arrs.dtype)

arr = arrs.astype(dtype=np.float64)
print(arr)
print(arr.dtype)
```
