```python
# matplotlib 

import pandas as pd
import matplotlib.pyplot as plt


x = [2, 3, 5, 6]
y= [4, 9, 25, 36]

plt.plot(x,y)
# plt.show()
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Practice Plot")
```

```python
df = pd.read_csv("mltdata.csv")


plt.plot(df["month"],df["sales"])
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales vs Month")
plt.show()
df.head(2)
```

```python
# 4. Customizing Line Plot

plt.plot(df["month"],df["sales"] , color='red' , marker='o' , linewidth=2, markersize=8)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales vs Month")
plt.show()
```

```python
# 5. Usage of Scatter Plot

from pyparsing import line


hour = df["study_hours"]
score = df["exam_score"]

plt.scatter(hour, score , color='orange')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Exam vs Study Relation")
plt.show()
```

```python
# 7. Histogram
# A histogram shows the frequency distribution of numerical data.

dep = df["department"]
salary = df["employee_salary" ]

plt.hist(salary , bins=8 , color='skyblue' , edgecolor='black' , histtype='bar')
plt.xlabel("Employee Salary")
plt.ylabel("Frequency")
plt.title("Distribution of Employee Salaries")
plt.show()
# plt.legend("hello")
```

```python
plt.pie(salary , labels=dep , colors=['lightcoral' , 'lightblue' , 'lightgreen' , 'lightsalmon' , 'lightyellow'] ,autopct='%5.1f%%' , startangle=90)
plt.title("Department-wise Salary Distribution")
plt.show()
```

```python
plt.plot()       # Line chart
plt.scatter()    # Scatter plot
plt.hist()       # Histogram
plt.pie()        # Pie chart
plt.title()      # Chart title
plt.xlabel()     # X-axis label
plt.ylabel()     # Y-axis label
plt.grid()       # Grid
plt.legend()     # Legend
plt.show()       # Display chart
```
