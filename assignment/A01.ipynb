Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.

Input Format: Space-separated integers

Output Format: A list of numbers satisfying the condition

Example: Input: 4 6 8 12 16 20 24 28

Output: [4, 12, 20, 28]



```python
lst=[4, 6, 8, 12, 16, 20, 24, 28]

res = list(filter(lambda x:  x%4 == 0 and x%8 !=0, lst))
print(res)
```

Write a function compute_ratio(x, y) that returns (x + y) / (x * y). Use exception handling to manage division by zero.
Use exception handling to manage:
Division by zero

Input Format:
Two values

Output Format:
Computed result OR error message


```python
def compute_ratio(x,y):
    if x * y == 0:
        try:
            return (x+y)/(x*y)
        except ZeroDivisionError:
            return "Error: Division by zero is not allowed."
    return (x+y)/(x*y)
compute_ratio(3,1)
```

Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.

Input Format:

A single line sentence

OutputFormat:
A list of strings



```python
find_vowel= input()
words = find_vowel.split()
result = []

for word in words :
    if word[0].lower() in "aeiou":

        result.append(word)
print(words)
print(result)
```

Write a Python program that reads a text file and finds the longest word.

Requirements:

Open and read data from a file (Example: data.txt) If multiple words have the same longest length, print any one of them

Input Format: A text file containing multiple lines of text

Output Format: The longest word

Example File Content (data.txt): Python is a wonderful language for machine learning.

Output: wonderful

```python
with open("file.txt" , "r") as file:
    text = file.read()
    # print(text)
words = text.split()
longest = words[0]

for word in words:
    # print(word)
    if len(word) > len(longest):
        longest = word
print(longest)
```

A word is called a Palindrome if reversing its letters results in the same word (ignoring case). Write a program to check whether a given string is a palindrome.

Input Format:
A single word string

Output Format:
"Palindrome" or "Not Palindrome"

Example:
Input:
Radar

Output:
Palindrome


```python
word = input("Write a word: ")
word = word.lower()

if word == word[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")

print(word[::-1])
```
