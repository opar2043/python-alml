```python
# You must run this code and use this dataset for all questions

import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

## Q4.
Create a new NumPy array where each employee’s age is adjusted based on:
Add 4 years if age is less than the average age
Add 2 years if age is greater than or equal to the average age
Do NOT modify the original array.
 
Print:
original array
updated array
average age used for condition


```python
avg_age = np.mean(emp_ages)
print("avg age: " ,avg_age)
copy_age = np.copy(emp_ages)
print("original array: " ,copy_age)
update_age = np.where(copy_age < avg_age, copy_age + 4, copy_age+ 2)
print("updated array: " ,update_age)
```

```python
# ===== Q5 ANSWER =====
# All Tech role salaries (Software Developer / Backend Developer) increase by 8%
import numpy as np

data = np.genfromtxt("staff_data.csv", delimiter=",", skip_header=1, dtype=str)
salaries = np.array(data[:, 2], dtype=float)
tech_roles = np.array(["Backend Developer", "Software Developer"])
is_tech = np.isin(data[:, 3], tech_roles)
tech_salaries = salaries[is_tech]

old_total = np.sum(salaries)
new_total = np.sum(salaries) + np.sum(tech_salaries * 0.08)
print("Q5 - Old total salary:", old_total)
print("Q5 - New total salary:", new_total)
print("Q5 - Increase in salary:", new_total - old_total)
```
