## Q7.
Filter the dataset to include only:
emp_id, emp_age, role
Then:
Select employees whose age is between 25 and 40
Sort them by emp_id
Display the last 3 rows of the result


```python
df_data = df[["emp_id", "emp_age" , "role"]]
filtered_data = df_data.loc[(df_data["emp_age"] >= 25) & (df_data["emp_age"] <= 40)]
sort_by_id = filtered_data.sort_values(["emp_id"] , ascending=True).tail(3)

sort_by_id
```

## Q10.
Using .loc:
Increase salary by 12% for "Software Developer" and "Backend Developer"
Increase salary by 5% for all other roles
Then:
Print the dataframe that includes role , previous_salary , new_salary , increase_of_salary

```python
df["previous_salary"]=df['emp_salary']
df["previous_salary"]

developer = df["role"].isin([
    "Software Developer",
    "Backend Developer"
])
df.loc[developer , "new_salary"] = df.loc[developer , "previous_salary"] * 1.12
df.loc[developer , "new_salary"] = df.loc[developer , "previous_salary"] * 1.05

df["increase_of_salary"] = df["new_salary"] - df["previous_salary"]
print(df[[
    "role",
    "previous_salary",
    "new_salary",
    "increase_of_salary"
]])
```

```python


df["previous_salary"] = df["emp_salary"]

developer = df["role"].isin([
    "Software Developer",
    "Backend Developer"
])


df.loc[developer, "new_salary"] = df.loc[developer, "previous_salary"] * 1.12


df.loc[~developer, "new_salary"] = df.loc[~developer, "previous_salary"] * 1.05


df["increase_of_salary"] = df["new_salary"] - df["previous_salary"]


print(df[[
    "role",
    "previous_salary",
    "new_salary",
    "increase_of_salary"
]])
```
