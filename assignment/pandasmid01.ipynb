## Q7.
Filter the dataset to include only:
emp_id, emp_age, role
Then:
Select employees whose age is between 25 and 40
Sort them by emp_id
Display the last 3 rows of the result


```python
df_data = df[["emp_id", "emp_age" , "role"]]
filtered_data = df_data.loc[(df_data["emp_age"] >= 25) & (df_data["emp_age"] <= 40)]
sort_by_id = filtered_data.sort_values(["emp_id"] , ascending=True).tail(3)

sort_by_id
```

## Q8.
Filter the DataFrame to show only employees working in the "HR" role.

```python
role = df[df["role"] == "HR Manager"]
print(role)
print("total HR employees:", len(role))
```
