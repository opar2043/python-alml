```python
#  transpose 
a = np.random.randint(1 ,10 , size =(10))
print(a)

txp = a.T
print("after transpose:")
print(txp)

split_a = np.array_split(a, 1)
print("after splitting:")
print(split_a)
```

```python
# faster for vectorization
x = np.array([ 10 , 8 , 30 , 100 ])
y = np.array([ 2 , 3 ,4  ,5 ])

add = np.add(x,y)
print(add)

sub = x-y
print(sub)

mul = x*y
print(mul)

div = x/y
print(div)

remainder = x % y
print(remainder)
```

```python
# mathematical function
# trigonometry
# sin , cos , tan takes values in radian
x = np.array([ 10 , 8 , 30 , 100 ])

sin_val = np.sin(x)
print(sin_val)
cos_val = np.cos(x)
print(cos_val)
tan_val = np.tan(x)
print(tan_val)

deg = np.deg2rad(x)
print(deg)

baseLog = np.log10(x)
baseLog2 = np.log2(x)
print(baseLog)
print(baseLog2)

sqt = np.sqrt(x)
print(sqt)

# prefix sum

cmul = np.cumulative_sum(x)
print(cmul)
```

```python
# comparison - >,< , >= ,<= , == , !=
import numpy as np
x = np.array([5, 50, 15, 20])
y = np.array([55, 15, 85, 30])

greater = x > y 
print(greater)

equal = x == y 
print(equal)

print(np.all(greater))
print(np.all(equal))
print(np.any(equal))
```

```python
#  Shorting 

x = np.array([5, 50, 15, 20])

sorted_x = np.sort(x)
print(sorted_x)

z= x.copy()
z.sort()
print(z)


# horizontal axis = 1

y = np.array([[2,1,9],[3,6,8]])
print(y)

hor_sort = np.sort(y , axis =1)
print(hor_sort)

vert_sort = np.sort(y , axis = 0)
print(vert_sort)
```

### searching

```python
x = np.array([5, 50, 15, 20])

idx = np.argsort(x)
idx_value = np.where( x > 15 )
print(idx)
print(idx_value)

arr = np.where(x > 8 , x , 0)
print(arr)
```

### Statical function

```python
# data load
data = np.genfromtxt('stddata.csv', delimiter=',' , skip_header=1)
print(data)
```

```python
math_marks = data[:, 0]
print(math_marks)

max_math  = np.max(math_marks)
print(max_math)
min_math  = np.min(math_marks)
print(min_math)
mean_math  = np.mean(math_marks)
print(mean_math)
median_math = np.median(math_marks)
print(median_math)

standard_math_marks = np.std(math_marks)
print(standard_math_marks)
```

### linear algenbra

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

dot = np.dot(A,B)
print(dot)

trx = np.trace(A)
print(trx)

rank_sq = np.linalg.matrix_rank(A)
print(rank_sq)
```
