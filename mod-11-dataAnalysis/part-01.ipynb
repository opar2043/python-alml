```python
import numpy as np

arr = np.random.randint(1,10 , size=(2,3))
print(arr)
print(arr.shape)
```

```python
b = arr.reshape(3,2)
print(b)
print(b.shape)

print("================")
flt = b.flatten()
print(flt)
clm_flt = np.ravel(b, order='C')
print(clm_flt)
```

```python
### Concatinate 
#? row wise concetanate korar shomoi column same thakbe
#? col wise concat korar shomoi row ta same thakte hobe


a = np.random.randint(1 ,10 , size =(2,3))
b = np.random.randint(20 ,30 , size =(2,3))
# print(a)
# print(b)

con_row=np.concatenate((a,b) , axis=0)  ## row wise
# print(con_row)
# print(con_row.shape)
con_col=np.concatenate((a,b) , axis=1)  ## column wise
print(con_col)
print(con_col.shape)
```

```python
#  transpose 
a = np.random.randint(1 ,10 , size =(10))
print(a)

txp = a.T
print("after transpose:")
print(txp)

split_a = np.array_split(a, 1)
print("after splitting:")
print(split_a)
```

```python
# faster for vectorization
x = np.array([ 10 , 8 , 30 , 100 ])
y = np.array([ 2 , 3 ,4  ,5 ])

add = np.add(x,y)
print(add)

sub = x-y
print(sub)

mul = x*y
print(mul)

div = x/y
print(div)

remainder = x % y
print(remainder)
```

```python
# mathematical function
# trigonometry
# sin , cos , tan takes values in radian
x = np.array([ 10 , 8 , 30 , 100 ])

sin_val = np.sin(x)
print(sin_val)
cos_val = np.cos(x)
print(cos_val)
tan_val = np.tan(x)
print(tan_val)

deg = np.deg2rad(x)
print(deg)

baseLog = np.log10(x)
baseLog2 = np.log2(x)
print(baseLog)
print(baseLog2)

sqt = np.sqrt(x)
print(sqt)

# prefix sum

cmul = np.cumulative_sum(x)
print(cmul)
```
