### 1 . Create a 1D, 2D, and 3D NumPy array and print their dimensions and shapes.


```python
import numpy as np

# Create a 1D array
arr1d = np.array([1, 2, 3, 4, 5])
print("1D Array:")
print(arr1d)
print("Dimensions:", arr1d.ndim)
print("Shape:", arr1d.shape)

# Create a 2D array
arr2d = np.array([[1, 2, 3], [4, 5, 6]])
print("\n2D Array:")
print(arr2d)
print("Dimensions:", arr2d.ndim)
print("Shape:", arr2d.shape)

# Create a 3D array
arr1 = np.array([1,2,3,4])
arr2 = np.array([[1,2,3], [4,5,6]])
arr3d = np.array([[[1,2], [3,4]], [[5,6], [7,8]]])

print(arr1.ndim)
print(arr1.shape)
print("=============")
print(arr2.ndim)
print(arr2.shape)
print("=============")
print(arr3d.ndim)
print(arr3d.shape)
```

4 . Create a 4×4 matrix from 0–15. Then , extract:
the first row

the last column
