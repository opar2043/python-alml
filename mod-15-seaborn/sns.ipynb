```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# import plotly.express as px
```

```python
emp = pd.read_csv('emp.csv')
emp.head()
```

```python
sns.lineplot(data = emp , x= "joining_year" , y = "salary" , errorbar=None) 
```

```python
sns.lineplot(data = emp , x= "age" , y = "salary" , errorbar=None , hue="gender") 
```

 ## Relplot

```python
sns.relplot(kind="line", data = emp , x="salary" , y="department" , errorbar=None , hue="gender")
```

```python
sns.relplot(kind="line", data = emp , y="salary" , x="experience" , errorbar=None , hue="gender")
```

```python
sns.scatterplot(data=emp , x="age" , y="salary" , hue="gender")
```

```python
sns.relplot( kind="scatter" ,data=emp , x="age" , y="salary" , hue="gender")
```

```python
sns.relplot( kind="line" ,data=emp , x="age" , y="salary" , hue="gender")
```

```python
sns.histplot(data=emp , y="salary" , x="department" , hue="department")
```

```python
sns.displot( kind="hist" , data=emp , x="salary"  , hue="gender" , bins=10)
```

```python
sns.kdeplot(data=emp , x="salary"  , hue="gender")
```

```python
sns.displot(kind="kde", data=emp, x="salary", hue="gender")
```

```python
sns.countplot(data=emp, x="department" , hue="gender")
```

```python
sns.countplot(data=emp, x="age" , hue="gender")
```

```python
# Bar plot
sns.barplot(data=emp, x="department", y="salary", hue="gender")
```

```python
# regplot
sns.regplot(data=emp , x="age" , y="salary")
```

```python
sns.pairplot(data=emp , vars=["age", "salary" , "joining_year" ] , hue="gender")
```
