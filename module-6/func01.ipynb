### iterator and generator 
#### Remember: iter() creates an iterator from an iterable, next() gets the next value, and yield produces values from a generator.

```python
num = iter([10,20,33,56])
print(type(num))
print(next(num))
print(next(num))
print(next(num))

help(iter)
```

```python
def count():
    for i in range(5):
        print(i)
count()        
```

### Lambda, Map, and Apply Function

```python
sqr = lambda x : x*x
print(sqr(5))
```

```python
num = [1, 2, 3, 4, 5]
res = list(map(lambda x : x + 2 , num))
print(res)

sqr = list(map(lambda y : y**2 , num))
print(sqr)
```
