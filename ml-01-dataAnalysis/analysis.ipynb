## categorical value analysis


```python
sns.countplot(data = df , x=df['Survived'])
```

```python
(df['Survived'].value_counts() / len(df) ) * 100
```

```python
values =df['Survived'].value_counts()
labels = df['Survived'].unique()
values
```

```python
plt.pie(values , labels=labels , autopct='%1.1f%%')
plt.title("Survivable Pie Chart")
plt.show()
```

```python
sns.countplot(data = df , x = 'Pclass')
```

```python
passengerClass = (df['Pclass'].value_counts() / len(df)) * 100
passengerClass
```

```python
plt.pie(passengerClass , labels=df['Pclass'].unique() , autopct='%1.1f%%')
plt.show()
```

```python
sns.histplot(df["Age"] , bins=40)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Person")
plt.show()
```
