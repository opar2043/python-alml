```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('emply.csv')
```

Read the CSV using Pandas.



Display the first 5 and last 5 rows.



Find the average, maximum and minimum salary.



Filter employees whose salary is greater than 50,000.



Find the average salary for each department using groupby().



Add a new column called bonus (10% of salary).



Add another column called new_salary.



Sort employees by salary in descending order.



Create a bar chart showing average salary by department.



Create a histogram showing salary distribution.



Create a scatter plot showing experience vs salary.

```python
df.head()
df.tail()
```

```python
salary = df['salary']
salary
avg_salary = salary.mean()
print("avg_salary:", avg_salary)
max_salary = salary.max()
print("max_salary:", max_salary)
min_salary = salary.min()
print("min_salary:", min_salary)
```

```python
filter_salary = df[df['salary'] > 50000]
filter_salary
```

```python
grp_salary = df.groupby('department')['salary'].mean()
grp_salary
```

```python
df[ 'bonus' ] = df["salary"] * 0.1
df.head(2)
```

```python
df["new_salary"] = df["salary"] + df["bonus"]
df.head(2)
```

```python
df.sort_values('salary' , ascending=False)
df.head()
```

```python
plt.bar(df["department"], df["salary"])
plt.xlabel("Department")
plt.ylabel("Salary")
plt.title("Salary Distribution by Department")
plt.show()
```

```python
plt.hist(df["salary"] , bins=20)
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")
plt.show()
```

```python
plt.scatter(df['salary'] , df["experience"])
plt.xlabel("Salary")
plt.ylabel("Experience")
plt.title("Salary vs Experience")
plt.show()
```
