```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("ecom.csv")
df.head(3)
```

```python
print(df.describe())
df.info()
```

```python
df.isnull()
df.duplicated()
```

```python
df["revenue"] = df["quantity"] * df["unit_price"]
df.head(3)
```

```python
total_revenue = df.groupby('product')['revenue'].sum()
print('Product-wise Total Revenue:')
print(total_revenue)
```

```python
total_revenue_category = df.groupby('category')['revenue'].sum()
print('Product-wise Total Revenue:')
print(total_revenue_category)
```

```python
top_selling = df['quantity'].max()
print(f"Top Selling Product Quantity: {df[df['quantity'] == top_selling]['product'].iloc[0]}")
```

```python
orders = df[df['revenue'] > 20000]
orders
```

```python
revenue = np.array(df['revenue'])
revenue

print("Total: " , np.sum(revenue))
print("Average: " , np.mean(revenue))
print("Minimum: " , np.min(revenue))
print("Maximum: " , np.max(revenue))
print("Standard Deviation: " , np.std(revenue))
```

```python
greater_avg = revenue > np.average(revenue)
print("Greater than Average:")
print(revenue[greater_avg])
```

```python
increase_revenue = revenue * 1.05 
print("Increased Revenue:", increase_revenue)
```
