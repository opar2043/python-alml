```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('student.csv')
print(df.head(2))
print("====================")
print(df.info())
print(df.describe())
print(df.shape)
print(df.isnull().sum())
```

```python
x1 = df['hours'].values
x2 = df['tests'].values
y = df['score'].values

# total sample
m = len(y)   

# mean values 
x1_mean = x1.mean()
x1_std = x1.std()
x2_mean = x2.mean()
x2_std = x2.std()

print(x1_mean)
print(x1_std)
```

```python
# x1 and x2 updated values
x1 = (x1 - x1_mean) / x1_std
x2 = (x2 - x2_mean) / x2_std
```

```python
# initialize parameter
w0 = 0 
w1 = 0 
w2 = 0

alpha = 0.05 
iteration = 1000
chat_history = []
```

```python
for i in range(iteration):
    ypred = w0 + w1*x1 + w2*x2
    error = ypred - y 

    cost = (1 / (2 * m))  *  np.sum(error ** 2)
    chat_history.append(cost)   # store only the cost (a single number)

    dw0 = (1/m) * np.sum(error)
    dw1 = (1/m) * np.sum(error * x1)
    dw2 = (1/m) * np.sum(error * x2)

    w0 = w0 - alpha* dw0
    w1 = w1 - alpha* dw1
    w2 = w2 - alpha* dw2

print("final weights: ")
print("dw0 = ", dw0)
print("dw1 = ", dw1)
print("dw2 = ", dw2)
print("w0 = ", w0)
print("w1 = ", w1)
print("w2 = ", w2)

print("Final Cost: " , chat_history[-1])
```

```python
plt.plot(chat_history)

plt.xlabel("iteration")
plt.ylabel("Cost (J)")

plt.title("Cost Reduction Using Gradient Descent")
plt.grid(True)
plt.show()
```
