```python
# 5. Separate Features and Target
x1 = df['hours'].values
x2 = df['tests'].values
y = df['score'].values

m = len(y)
```

```python
# 6. Feature Scaling
x1_mean = np.mean(x1)
x2_mean = np.mean(x2)

x1_std = np.std(x1)
x2_std = np.std(x2)

x1 = (x1 - x1_mean) / x1_std
x2 = (x2 - x2_mean) / x2_std
```

```python
# 7. Initialize Weights and Parameters
w0 = 0
w1= 0
w2 = 0

alpha = 0.05 
itr = 100

chat_history = []
```

```python
# 8. Implement Gradient Descent
y_pred = w0 + w1*x1 + w2*x2

# B. Calculate Error
error = y_pred - y
```

```python
# C. Calculate Cost
cost = (1 / (2 * m)) * np.sum(error ** 2)
chat_history.append(cost)
```

```python
# D. Calculate Gradients
dw0 = (1 / m) * np.sum(error)
dw1 = (1 / m) * np.sum(error * x1)
dw2 = (1 / m) * np.sum(error * x2)
```

```python
# E. Update Weights
w0 = w0 - alpha * dw0
w1 = w1 - alpha * dw1
w2 = w2 - alpha * dw2
```

```python
plt.plot(chat_history)
plt.xlabel('Iterations')
plt.ylabel("cost")
plt.title("Cost Function")
plt.show()
```

```python
# 10. Display Regression Equation
print(f"Regression Equation: y = {w0:.2f} + {w1:.2f}*x1 + {w2:.2f}*x2")
```

```python
# actual and predicted scores
y_pred = w0 + w1 * x1 + w2*x2 
result = pd.DataFrame({
    "actual_value" : y,
    "predicted_value" : y_pred
})

df.head(5)
```
