```python
import numpy as np

names = np.array([
    "John", "Sarah", "David", "Emma",
    "Michael", "Olivia", "James", "Sophia"
])

math = np.array([85, 90, 78, 92, 65, 88, 76, 95])
physics = np.array([80, 85, 82, 95, 70, 90, 75, 98])
chemistry = np.array([88, 92, 75, 89, 68, 85, 80, 96])
```

Problem 2: Student Performance Analysis (NumPy)
Scenario: You are a Data Analyst at an educational institution. You need to analyze students' examination results using NumPy.

Calculate each student's total marks.



Calculate each student's average marks.



Find the highest and lowest average marks.



Find the student with the highest total marks.



Find the class average for each subject.



Find students who scored more than 80 in all subjects.



Give every student a 5% bonus on their total marks.



Find the number of students whose average is greater than 85.



Sort students by average marks in descending order.



Create a 2D NumPy array containing all subject marks.

```python
total = math + physics + chemistry
total
```

```python
avg = total / 3
print(avg)
```

```python
print("max avg:", avg.max())
print("min avg:", np.min(avg))
```

```python
print("highest Ttal marks:",  np.max(total))
```

```python
print("math avg:" , math.mean())
print("physics avg:" , physics.mean())
print("chemistry avg:" , chemistry.mean())
```

```python
condition = (math > 80) & (physics > 80) & (chemistry > 80)
std_name = names[condition]
print(std_name)
```

```python
bonus_total = total * 1.05
bonus_total
```

```python
num_std = avg > 85
print("Total 85+ students:", num_std.sum())

print("Descending avg order: " , avg[np.argsort(avg)[::-1]])
```

```python
marks = np.column_stack([math, physics, chemistry])
print("2D array")
print(marks)
```
